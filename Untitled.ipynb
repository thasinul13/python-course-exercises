```python
movie = "It's a Mega Blockbuster!"

movie.split()          # splits on whitespace by default
movie.split("'")       # splits on a custom separator instead — here, the apostrophe

word_list = movie.split()
word_list

' '.join(word_list)    # glues the list back into a string, with a space between each word

movie.lower().replace("'", '').rstrip('!')   # chained: lowercase, remove the apostrophe, strip trailing !
```

```python
(not False) * 10        # True * 10 -> 10

'word' == 'word'        # True
'word' == 'WORD'        # False — case-sensitive
'word' == 'WORD'.lower()  # True — .lower() makes both sides match

'word' != 'word'        # False

len('word') > len('word')   # False — equal lengths, not greater
len('word') >= len('word')  # True — equal counts, since >= allows equal
```

```python
len('word') > len('word')     # False — equal, not greater
len('word') >= len('word')    # True — equal counts as >=

len('words') > len('word')    # True — 5 > 4
len('word') < len('word')     # False — equal, not less
len('word') <= len('word')    # True — equal counts as <=

'or' in 'word'    # True — 'or' appears inside 'word'
'nor' in 'word'   # False — 'nor' isn't a substring of 'word'
```

```python
price = 499.99
budget = 500
inventory = 0

if budget > price:
    if inventory > 0:
        print('You can afford this and we have it in stock!')
    else:  # equivalent to if inventory <= 0
        print("You can afford this but it's out of stock.")
else:
    print(f'Unfortunately, this board costs more than ${budget}.')
```
