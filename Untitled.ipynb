```python
(not False) * 10        # True * 10 -> 10

'word' == 'word'        # True
'word' == 'WORD'        # False — case-sensitive
'word' == 'WORD'.lower()  # True — .lower() makes both sides match

'word' != 'word'        # False

len('word') > len('word')   # False — equal lengths, not greater
len('word') >= len('word')  # True — equal counts, since >= allows equal
```

```python
len('word') > len('word')     # False — equal, not greater
len('word') >= len('word')    # True — equal counts as >=

len('words') > len('word')    # True — 5 > 4
len('word') < len('word')     # False — equal, not less
len('word') <= len('word')    # True — equal counts as <=

'or' in 'word'    # True — 'or' appears inside 'word'
'nor' in 'word'   # False — 'nor' isn't a substring of 'word'
```

```python
customer_list = [
    'C00001', 'C00003', 'C00004', 'C00005', 'C00006',
    'C00007', 'C00018', 'C00010', 'C00013', 'C00014',
    'C00015', 'C00016', 'C00029'
]

# Did C00009 make a purchase? Return 99.99 if yes, 0.0 if no
('C00009' in customer_list) * 99.99   # -> 0.0

# 1. Fifth and sixth customer
fifth_sixth = customer_list[4:6]
fifth_sixth   # -> ['C00006', 'C00007']

# 2. Every third customer (1st, 4th, 7th...)
every_third = customer_list[::3]
every_third   # -> ['C00001', 'C00005', 'C00018', 'C00014', 'C00029']

# 3. Last two customers
last_two = customer_list[-2:]
last_two   # -> ['C00016', 'C00029']
```
