```python
movie = "It's a Mega Blockbuster!"

movie.split()          # splits on whitespace by default
movie.split("'")       # splits on a custom separator instead — here, the apostrophe

word_list = movie.split()
word_list

' '.join(word_list)    # glues the list back into a string, with a space between each word

movie.lower().replace("'", '').rstrip('!')   # chained: lowercase, remove the apostrophe, strip trailing !
```

```python
text1 = '            Your friend Mark'
text2 = 'was'
text3 = 'having'
text4 = 'a great day'
text5 = 'on the mountain'
```

```python
full_text = text1 + ' ' + text2 + ' ' + text3 + ' ' + text4 + ' ' + text5
full_text
```

```python
fixed_text = full_text.strip().lower()
fixed_text
```

```python
new_text = fixed_text.replace('on the mountain', 'at the ski shop')
new_text
```

```python
word_count = new_text.count(' ') + 1
word_count
```
